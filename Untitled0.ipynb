```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="pie",
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",

figsize=(11,8),
color=["pink"])

plt.title("Patron Registrations by Year")
plt.show()
```

```python
month_order = [
"January", "February", "March", "April", "May", "June",
"July", "August", "September", "October", "November", "December"]

df["Circulation Active Month"].value_counts().plot(
    kind="bar",

figsize=(11,8),
color=["pink"])

plt.title("Library Circulation Activity by Month")
plt.ylabel("#of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
df.isna().sum()
```

```python
df["Circulation Active Month Cleaned"] = df["Circulation Active Month"].fillna('Not Captured')
```

```python
month_order = [
"January", "February", "March", "April", "May", "June",
"July", "August", "September", "October", "November", "December"]

month_counts = df["Circulation Active Month Cleaned"].value_counts().reindex(month_order)

month_counts.plot(
    kind="bar",

figsize=(11,8),
color=["pink"])

plt.title("Library Circulation Activity by Month")
plt.ylabel("#of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December", "Not Captured"]
month_counts = df["Circulation Active Month Cleaned"].value_counts().reindex(month_order)

month_counts.plot(
    kind="bar",
    color="pink",
    edgecolor="black",
    figsize=(10,6)
)

plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python

value_counts = df["Patron Type Definition"].value_counts()

s_temp = df["Patron Type Definition"].sort_values(ascending=False)
not_top_five = len(s_temp)-5
not_top_five_sum = s_temp.tail(not_top_five).sum()
s_top = s_temp.head(5)
s_top[5] = not_top_five_sum

value_counts.plot(
    kind="pie",
    figsize=(10,6)
)

plt.title("Library Patron Age Groups", fontsize=16, weight="bold")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
